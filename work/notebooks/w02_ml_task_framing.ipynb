## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

**Why ML:** The pattern is too messy and multi-dimensional for an if-statement. If we tried to write rules like `if impressions > 1000 and clicks == 0: 'hidden gem'`, we would quickly fail because the data has extreme variance (median impressions 731, max 517,715) and multiple tangled signals. Hardcoded thresholds will either miss thousands of pages or create dozens of overlapping, unmanageable categories. ML (like K-Means) scales these variables and finds the natural boundaries across all dimensions simultaneously.
