## 3. Quick look at the data (2-3 real numbers)

The data shows massive variance that simple rules will struggle to categorize. For example, while the median page receives 731 impressions, the top page receives 517,715. Furthermore, there are 5,831 pages sitting on page one of Google that receive zero clicks. Clustering is necessary to group these wildly different performance behaviors into structured archetypes.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

# 1. Show the extreme variance in traffic
max_impressions = df['impressions_90d'].max()
median_impressions = df['impressions_90d'].median()

# 2. Show that some pages rank high but get ignored (perfect for finding 'hidden gem' archetypes)
high_rank_low_clicks = df[(df['avg_position'] <= 10) & (df['clicks_90d'] == 0)].shape[0]

print(f"Data Snapshot:")
print(f"1. Extreme Variance: The median page gets {median_impressions:,.0f} impressions, but the top page gets {max_impressions:,.0f}.")
print(f"2. Hidden Patterns: There are {high_rank_low_clicks:,} pages that rank on page one (position <= 10) but received exactly 0 clicks.")
```
